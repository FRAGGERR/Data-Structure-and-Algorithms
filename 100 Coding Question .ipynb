`arr.sort()` (In-place method)<br>
Type: <b>**Method of list objects**<b>. <br>
Modification: Modifies the original **list in-place** (<u>no new list is created</u>). <br>
Return value: It does not return anything (None). It only sorts the list on which it is called. <br>
Use case: Use arr.sort() when you want to sort the list and do not need to preserve the original order of the elements. <br>

`sorted(hardik)` (Built-in function)<br>
Type: **Built-in Python function**.<br>
Modification: <u>It does not modify the original iterable</u> (string, list, etc.).<br>
Return value: It returns a new sorted list. The original list (or iterable) remains unchanged.<br>
Use case: Use sorted() when you need a sorted version of an iterable but want to keep the original order intact.<br>

## Q.10 How to calculate the number of vowels and consonants in a string?

```python
vowels = ['a', 'e', 'i', 'o', 'u', 'A','E', 'I', 'O', 'U']

string = '@Janenie Janakiraman'

def vowels_consonants (string):
    vowels_count = 0
    consonant_count = 0
    for i in string:
        if i in vowels:
            vowels_count += 1
        elif i.isalpha(): # Check if the character is a letter (to avoid counting spaces):
            consonant_count += 1
    return vowels_count, consonant_count


print(vowels_consonants(string))
```

## Q.12 Compute the first Ten Fibonacci numbers.

```python
n = int(input())

def fibonacci_series(n):
    a, b = 0, 1
    series = []
    for i in range (n):
        series.append(a)
        next = a + b
        a = b
        b = next
    return series


print(fibonacci_series(n))
```

## Q.14 How to search for a specific value in a linked list?

```python

# A Linked List Node
class Node:
  
      # Constructor to intialize a node with data
    def __init__(self, new_data):
        self.data = new_data
        self.next = None

# Checks whether key is present in linked list
def search_key(head, key):
  
    # Initialize curr with the head of linked list
    curr = head

    # Iterate over all the nodes
    while curr is not None:

        # If the current node's value is equal to key,
        # return true
        if curr.data == key:
            return True

        # Move to the next node
        curr = curr.next

    # If there is no node with value as key, return false
    return False

# Driver code
if __name__ == "__main__":

    # Create a hard-coded linked list:
    # 14 -> 21 -> 13 -> 30 -> 10
    head = Node(14)
    head.next = Node(21)
    head.next.next = Node(13)
    head.next.next.next = Node(30)
    head.next.next.next.next = Node(10)

    # Key to search in the linked list
    key = 14

    if search_key(head, key):
        print("Yes")
    else:
        print("No")
```

*    ales/seal, 
*    toga/goat,
*    hear/hare,

*    paws/wasp,
*    flow/wolf

```python
# to check that if the given string is an anagram or not, we are going to use the below function
# By SORTING the LIST

def anagramBySorting(string1, string2):
    string1, string2 = string1.replace(' ', '').lower(), string2.replace(' ', '').lower()
    if (sorted(string1) == sorted(string2)):
        return True

    return False

string1 = 'the eyes'
string2 = 'they see'

print(anagramBySorting(string1, string2))
```

## Q.24 Reverse the words in a sentence.

```python
def reverseStringWordWise(string):
    # write your code logic !!!
    words = string.split()
    reverse_words = []
    for word in reversed(words):
        reverse_words.append(word)
        reversed_sentence = " ".join(reverse_words)
    return reversed_sentence

string = 'hardik is the best coder in this world'
print(reverseStringWordWise(string))
```

```python
def reversethestring(string):
    words = string.split()
    n = len(words)
    output_list = []
    # start = n - 1, ends = -1(why -1 cause if we use 0 then the last element will be 1,
    #                          so to go to the complete end of the string we must have to use -1)
    for i in range(n-1, -1, -1):
        output_list.append(words[i])
        reversed_string = ' '.join(output_list)
    return reversed_string

string = 'leetcode is the shit platform'
print(reversethestring(string)) #function calling
```

## Q.25  Encode the Message

```python
def encode(message):
    string = []
    n = len(message)

    i = 0

    while(i<n):
        temp = message[i]
        count = 1

        for j in range(i + 1, n):
            if (message[j] == temp):
                count += 1
            else:
                break
                
        string.append(temp)
        string.append(str(count))

        i += count

    return "".join(string)


message = "aaabbbaaabbhhhardd"

print(encode(message))
```

## Q.26 Maximum Subarray Sum

```python
def maxSubarraySum(arr) :
    max_sum = 0
    curr_sum = 0

    for num in arr:
        curr_sum += num
        if (curr_sum > max_sum):
            max_sum = curr_sum
        if (curr_sum < 0):
            curr_sum = 0
    return max_sum

arr = [1,2,7,-4,3,2,-10,9,1]
print(maxSubarraySum(arr))
```

## Q.27 Given an array/list 'ARR' of integers and a position ‘M’. You have to reverse the array after that position.

```python
def reverseArray(arr, m):
    # Write your code here.
    length_arr = len(arr)
    new_arr = []
    for j in range(0, m+1):
        new_arr.append(arr[j])

    for i in range(length_arr-1,m,-1):
        new_arr.append(arr[i])
    return new_arr
```

## Q.29 Shortest Word Distance (243)

```python
string = ['hardik', 'likes', 'to', 'code', 'in', 'jupyter', 'Notebook']
word1 = 'hardik'
word2 = 'in'
def shortest_word_distance(string, word1, word2): # 1. taking the inputs
    min_distance = float('inf') # 2: assigning the mini distance positive infinity
    index1, index2 = -1, -1 # 3: creating variables for count the index value of word1 and word2 
    for i in range(len(string)): #4: iterate throung out the string and check of word1 and word2
        if string[i] == word1:
            index1 = i
        if string[i] == word2:
            index2 = i
        if (index1 != -1 and index2 != -1): #5: if there are no changes in the indexs -- Break it here
            min_distance = min(min_distance, abs(index2 - index1)) #6: calculate the mini btw the infinity and the difference of index1, index2
    return min_distance
print(shortest_word_distance(string, word1, word2))
```

```python
words = ["Hello","Alaska","Dad","Peace"]
def whichrowfromkeyboard(words):
    
    first_row = set('qwertyuio')  #1: creating the predefinied variables
    second_row = set('asdfghjkl')
    third_row = set('zxcvbnm')

    result = []  #2: empty list to append the results
    
    for word in words: #3: main result
        lower_word = set(word.lower()) #4: lower_case the words, cause the test cases are case senative words
        if (lower_word.issubset(first_row) or lower_word.issubset(second_row) or lower_word.issubset(third_row)): #5: Particular word is the subset of the any those rows, append that word
            result.append(word)
    return result

print(whichrowfromkeyboard(words))
```

## Q.32 Max Consecutive Ones (485)


```python
nums = [1, 1, 0, 0, 1]
def findMaxConsecutiveOnes(nums: List[int]) -> int:
    max_count = 0
    count = 0
    for num in nums:
        if (num == 0):
            count += 1
            max_count = max(max_count, count)
        else:
            count = 0
    return max_count
print(findMaxConsecutiveOnes(nums))
```
