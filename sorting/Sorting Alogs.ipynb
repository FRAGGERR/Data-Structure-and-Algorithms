```python
def bubbleSort(arr):
    n = len(arr)-1
    for passes in range(0, n+1): # Loop for the passes
        for j in range(0, n): # Loop for the shifting the elements
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

unsortedArr = [12, 25, 11, 34, 90, 22]
sortedArr = bubbleSort(unsortedArr)
print(sortedArr)
```

### Optimized way to do the above Bubble sorting

```python
def bubbleSort(arr):
    n = len(arr)-1
    for passes in range(0, n+1): 
        for j in range(0, n - passes): # Subtracting by Passes will always decrese the iteration.
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

unsortedArr = [12, 25, 11, 34, 90, 22]
sortedArr = bubbleSort(unsortedArr)
print(sortedArr)
```
